```python
import numpy as np
import json
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.logspace(-1, 5, 20)
```

```python
with open('training_regression_losses_21_04.txt', 'r') as f:
     all_losses = json.loads(f.read())

configs = np.array(list(all_losses.keys()))
losses_array = np.array([list(losses.values()) for losses in all_losses.values()])

mean_loss = np.mean(losses_array, axis=1)
P5, Q1, median_loss, Q3, P95 = np.quantile(losses_array, [0.05, 0.25, 0.5, 0.75, 0.95], axis=1)

best_mean_indices = np.argsort(mean_loss)[:5]
best_mean_models = configs[best_mean_indices]
best_mean_losses = mean_loss[best_mean_indices]

best_Q1_indices = np.argsort(Q1)[:5]
best_Q1_models = configs[best_Q1_indices]
best_Q1_losses = Q1[best_Q1_indices]

best_median_indices = np.argsort(median_loss)[:5]
best_median_models = configs[best_median_indices]
best_median_losses = median_loss[best_median_indices]

best_Q3_indices = np.argsort(Q3)[:5]
best_Q3_models = configs[best_Q3_indices]
best_Q3_losses = Q3[best_Q3_indices]

best_mean_index = best_mean_indices[0]
best_mean_model = configs[best_mean_index]
best_mean_model_losses = losses_array[best_mean_index]
best_mean_model_stats = {'mean': mean_loss[best_mean_index],
                         'P5': P5[best_mean_index],
                         'Q1': Q1[best_mean_index],
                         'median': median_loss[best_mean_index],
                         'Q3': Q3[best_mean_index],
                         'P95': P95[best_mean_index]}

best_median_index = best_median_indices[0]
best_median_model = configs[best_median_index]
best_median_model_losses = losses_array[best_median_index]
best_median_model_stats = {'mean': mean_loss[best_median_index],
                           'P5': P5[best_median_index],
                           'Q1': Q1[best_median_index],
                           'median': median_loss[best_median_index],
                           'Q3': Q3[best_median_index],
                           'P95': P95[best_median_index]}
```

```python
best_mean_losses
```

```python
sns.barplot(x=best_mean_models, y=best_mean_losses, color='blue', label='Mean Losses')
plt.xticks(rotation=45)
plt.ylabel('Loss')
plt.title('Best five mean Losses')
```

```python
sns.barplot(x=best_median_models, y=best_median_losses, color='orange', label='Median Losses')
plt.xticks(rotation=45)
plt.ylabel('Loss')
plt.title('Best five median Losses')
```

```python
sns.barplot(x=best_mean_model_stats.keys(), y=list(best_mean_model_stats.values()), color='blue', label='Best Mean Model Stats')
plt.xticks(rotation=45)
plt.title(f'Best Mean Model Stats ({best_mean_model})')
```

```python
sns.barplot(x=best_median_model_stats.keys(), y=list(best_median_model_stats.values()), color='orange', label='Best Median Model Stats')
plt.xticks(rotation=45)
plt.title(f'Best Median Model Stats {best_median_model}')
```
